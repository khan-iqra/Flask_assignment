Q4. Create /welcome route and / route

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def company_info():
    return """
    <h2>Company Details</h2>
    Company Name: ABC Corporation<br>
    Location: India<br>
    Contact Detail: 999-999-9999
    """

@app.route('/welcome')
def welcome():
    return "Welcome to ABC Corporation"

if __name__ == '__main__':
    app.run(debug=True)
```

```python
from flask import Flask, url_for

app = Flask(__name__)

@app.route('/')
def home():
    return "Home Page"

@app.route('/about')
def about():
    return "About Page"

with app.test_request_context():
    print(url_for('home'))
    print(url_for('about'))
```
